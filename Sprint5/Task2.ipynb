<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        // Присваиваем делегату метод DisplayInfo
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

// Делегаты
public delegate void DisplayDelegate();
public delegate void TransportHandler(string message);

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    // Событие: оповещает подписчиков об обновлении модели
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Модель изменена на {newModel}");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет багажник: {HasCarrier}");
    }
}
````

````
using System;
using System.Collections.Generic;
using System.Linq;

var myCar = new Car("Camry", "Toyota", 4);
var myBike = new Bike("Escape 3", "Giant", true);
var myCar2 = new Car("Model 3", "Tesla", 4);

// ===== 1. Делегаты =====
Console.WriteLine("=== Делегаты ===");
DisplayDelegate showCar = myCar.DisplayInfo;
DisplayDelegate showBike = myBike.DisplayInfo;

showCar();
showBike();
Console.WriteLine();

// Мультикаст-делегат: вызывает сразу несколько методов
DisplayDelegate showAll = myCar.DisplayInfo;
showAll += myBike.DisplayInfo;
showAll += myCar2.DisplayInfo;
Console.WriteLine("Вызов мультикаст-делегата:");
showAll();
Console.WriteLine();

// ===== 2. Лямбда-выражения =====
Console.WriteLine("=== Лямбда-выражения ===");
List<Transport> transports = new List<Transport>
{
    myCar,
    myBike,
    myCar2,
    new Car("Golf", "Volkswagen", 5)
};

// Фильтрация: только автомобили с 4 дверями
var fourDoorCars = transports.OfType<Car>().Where(c => c.NumberOfDoors == 4);
Console.WriteLine("Автомобили с 4 дверями:");
foreach (var item in fourDoorCars)
{
    item.DisplayInfo();
    Console.WriteLine();
}

// Проекция: список моделей, отсортированный по названию
var models = transports.OrderBy(t => t.Model).Select(t => t.Model);
Console.WriteLine("Модели по алфавиту: " + string.Join(", ", models));

// Func и Action с лямбдой
Func<Transport, string> describe = t => $"{t.Manufacturer} {t.Model}";
Action<string> print = text => Console.WriteLine(text);

foreach (var item in transports)
{
    print(describe(item));
}
Console.WriteLine();

// ===== 3. События =====
Console.WriteLine("=== События ===");
var myTransport = new Transport("Old Model", "Generic");

// Подписка лямбда-выражением
myTransport.TransportUpdated += message => Console.WriteLine($"[Подписчик 1] {message}");

// Подписка именованным обработчиком
TransportHandler logger = message => Console.WriteLine($"[Подписчик 2 - лог] {message}");
myTransport.TransportUpdated += logger;

myTransport.UpdateModel("New Model");   // сработают оба подписчика
Console.WriteLine();

// Отписка второго подписчика
myTransport.TransportUpdated -= logger;
myTransport.UpdateModel("Newest Model"); // сработает только первый
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) с применением  делегатов, лямда-выражений и событий.

````
using System;
using System.Collections.Generic;
using System.Linq;

// ===== Делегаты =====
public delegate void SoundDelegate();
public delegate void AnimalEventHandler(string message);

// ===== Базовый класс =====
public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }

    // Событие: оповещает подписчиков об изменении состояния животного
    public event AnimalEventHandler StateChanged;

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}");
    }

    public void Feed()
    {
        StateChanged?.Invoke($"{Name} покормлен(а)");
    }

    public void Birthday()
    {
        Age++;
        StateChanged?.Invoke($"У {Name} день рождения! Теперь возраст: {Age}");
    }
}

// ===== Производные классы =====
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, bool isIndoor) : base(name, age)
    {
        IsIndoor = isIndoor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor}");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }

    public Bird(string name, int age, bool canFly) : base(name, age)
    {
        CanFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Чирик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, int age, string waterType) : base(name, age)
    {
        WaterType = waterType;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} молчит... (рыбы не издают звуков)");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }
}

// Вспомогательный класс: метод принимает делегат как параметр
public class SoundPlayer
{
    public static void PlayTimes(SoundDelegate sound, int times)
    {
        for (int i = 0; i < times; i++)
        {
            sound();
        }
    }
}
````

````

var dog = new Dog("Рекс", 3, "Овчарка");
var cat = new Cat("Мурка", 2, true);
var bird = new Bird("Кеша", 1, true);
var fish = new Fish("Немо", 1, "Морская");

List<Animal> zoo = new List<Animal> { dog, cat, bird, fish };

// ===== 1. Делегаты =====
Console.WriteLine("=== Делегаты ===");
SoundDelegate dogSound = dog.MakeSound;
dogSound();
Console.WriteLine();

// Мультикаст-делегат: все животные подают голос
SoundDelegate chorus = dog.MakeSound;
chorus += cat.MakeSound;
chorus += bird.MakeSound;
chorus += fish.MakeSound;
Console.WriteLine("Хор животных:");
chorus();
Console.WriteLine();

// Удаляем метод из делегата
chorus -= fish.MakeSound;
Console.WriteLine("Хор без рыбки:");
chorus();
Console.WriteLine();

// Делегат как параметр метода
Console.WriteLine("Кошка мяукает 2 раза:");
SoundPlayer.PlayTimes(cat.MakeSound, 2);
Console.WriteLine();

// ===== 2. Лямбда-выражения =====
Console.WriteLine("=== Лямбда-выражения ===");

// Func: преобразует животное в строку
Func<Animal, string> describe = x => $"{x.GetType().Name}: {x.Name}, {x.Age} лет";
foreach (var item in zoo)
{
    Console.WriteLine(describe(item));
}
Console.WriteLine();

// Фильтрация
var olderAnimals = zoo.Where(x => x.Age >= 2).ToList();
Console.WriteLine("Животные старше 1 года: " + string.Join(", ", olderAnimals.Select(x => x.Name)));

// Сортировка по имени
var sortedNames = zoo.OrderBy(x => x.Name).Select(x => x.Name);
Console.WriteLine("По алфавиту: " + string.Join(", ", sortedNames));

// Агрегация
double averageAge = zoo.Average(x => x.Age);
Console.WriteLine($"Средний возраст: {averageAge:F1}");
Console.WriteLine();

// Action с лямбдой и ForEach
Console.WriteLine("Информация обо всех животных:");
zoo.ForEach(x =>
{
    x.DisplayInfo();
    Console.WriteLine();
});

// ===== 3. События =====
Console.WriteLine("=== События ===");

// Обработчик-лог для всех животных
AnimalEventHandler logger = message => Console.WriteLine($"[Лог] {message}");
foreach (var pet in zoo)
{
    pet.StateChanged += logger;
}

dog.Feed();
cat.Birthday();
Console.WriteLine();

// Дополнительный подписчик только на собаку
dog.StateChanged += message => Console.WriteLine($"[Хозяин Рекса] Получено уведомление: {message}");
Console.WriteLine("Теперь у Рекса два подписчика:");
dog.Feed();
Console.WriteLine();

// Отписываем лог от всех животных
foreach (var pet in zoo)
{
    pet.StateChanged -= logger;
}
Console.WriteLine("После отписки лога у Рекса остался только хозяин:");
dog.Feed();
cat.Feed(); // у кошки подписчиков нет, ничего не выведется
````
