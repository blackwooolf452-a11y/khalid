#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

В рамках данного проекта реализуется система управления задачами внутри проекта на языке C#.
Создан базовый класс `Task`, представляющий задачу с идентификатором, названием и приоритетом.
На его основе разработаны три производных класса, демонстрирующих принципы **наследования** и **полиморфизма**:

1. **DelegateTask (Делегатская задача)** — содержит дополнительные атрибуты: дата выполнения (`DueDate`), кто делегировал (`DelegatedBy`), отдел (`Department`), оценка в часах (`EstimatedHours`). Метод `MarkAsComplete()` переопределён для включения даты выполнения в сообщение о завершении.
2. **TeamTask (Командная задача)** — содержит дополнительные атрибуты: название команды (`TeamName`), размер команды (`TeamSize`), тимлид (`TeamLead`), список участников (`TeamMembers`). Метод `ReassignTo()` переопределён для указания нового члена команды.
3. **ResearchTask (Исследовательская задача)** — содержит дополнительные атрибуты: источник данных (`DataSource`), область исследования (`ResearchField`), дата начала (`StartDate`), список публикаций (`Publications`). Метод `GetTaskDetails()` переопределён для отображения источников данных.

#### Дополнительное задание
К существующим классам (базовому и производным) добавлены 3–4 атрибута и метода.
Реализованы три типа наследования:
- **Простое наследование**: `DelegateTask : Task`, `TeamTask : Task`.
- **Сложное (многоуровневое) наследование**: `ResearchTask` наследует `Task` и переопределяет цепочку методов, а также реализует интерфейсы.
- **Множественное наследование**: в C# реализуется через интерфейсы — `ResearchTask : Task, IReportable, IPrioritizable`.

```
using System;
using System.Collections.Generic;

// ============================================================
// 1. ИНТЕРФЕЙСЫ — для реализации множественного наследования
// ============================================================
public interface IReportable
{
    string GenerateReport();
}

public interface IPrioritizable
{
    void ChangePriority(string newPriority);
    string GetPriorityLevel();
}

// ============================================================
// 2. БАЗОВЫЙ КЛАСС Task
// ============================================================
public class Task
{
    // Атрибуты базового класса
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public bool IsCompleted { get; protected set; }
    public string AssignedTo { get; protected set; }

    public Task(int taskId, string taskName, string priority, string assignedTo)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = assignedTo;
        IsCompleted = false;
    }

    // Методы базового класса
    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[БАЗОВЫЙ] Задача '{TaskName}' (ID: {TaskId}) отмечена как выполненная.");
    }

    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | " +
               $"Исполнитель: {AssignedTo} | Статус: {(IsCompleted ? "Выполнена" : "В работе")}";
    }

    public virtual void ReassignTo(string newMember)
    {
        string old = AssignedTo;
        AssignedTo = newMember;
        Console.WriteLine($"[БАЗОВЫЙ] Задача '{TaskName}' переназначена с {old} на {newMember}.");
    }
}

// ============================================================
// 3. DelegateTask — ПРОСТОЕ НАСЛЕДОВАНИЕ + 4 атрибута, 2 метода
// ============================================================
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public string DelegatedBy { get; set; }
    public string Department { get; set; }
    public int EstimatedHours { get; set; }

    public DelegateTask(int id, string name, string priority, string assignedTo,
                        DateTime dueDate, string delegatedBy, string department, int hours)
        : base(id, name, priority, assignedTo)
    {
        DueDate = dueDate;
        DelegatedBy = delegatedBy;
        Department = department;
        EstimatedHours = hours;
    }

    // Переопределение MarkAsComplete (полиморфизм)
    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[DELEGATE] Задача '{TaskName}' завершена! " +
                          $"Срок выполнения был: {DueDate:dd.MM.yyyy}. " +
                          $"Отдел: {Department}, делегировал: {DelegatedBy}.");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $" | Срок: {DueDate:dd.MM.yyyy} | Отдел: {Department} | " +
               $"Делегировал: {DelegatedBy} | Часы: {EstimatedHours}";
    }

    // Новые методы
    public bool IsOverdue()
    {
        return !IsCompleted && DateTime.Now > DueDate;
    }

    public void ExtendDeadline(int days)
    {
        DueDate = DueDate.AddDays(days);
        Console.WriteLine($"Срок задачи '{TaskName}' продлён на {days} дней. Новый срок: {DueDate:dd.MM.yyyy}");
    }
}

// ============================================================
// 4. TeamTask — ПРОСТОЕ НАСЛЕДОВАНИЕ + 4 атрибута, 2 метода
// ============================================================
public class TeamTask : Task
{
    public string TeamName { get; set; }
    public int TeamSize { get; set; }
    public string TeamLead { get; set; }
    public List<string> TeamMembers { get; set; }

    public TeamTask(int id, string name, string priority, string assignedTo,
                    string teamName, int teamSize, string teamLead)
        : base(id, name, priority, assignedTo)
    {
        TeamName = teamName;
        TeamSize = teamSize;
        TeamLead = teamLead;
        TeamMembers = new List<string> { teamLead };
    }

    // Переопределение ReassignTo (полиморфизм)
    public override void ReassignTo(string newMember)
    {
        string old = AssignedTo;
        AssignedTo = newMember;
        if (!TeamMembers.Contains(newMember))
            TeamMembers.Add(newMember);

        Console.WriteLine($"[TEAM] Задача '{TaskName}' переназначена в команде '{TeamName}': " +
                          $"с {old} на {newMember}. Тимлид: {TeamLead}.");
    }

    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $" | Команда: {TeamName} | Размер: {TeamSize} | " +
               $"Тимлид: {TeamLead} | Участники: {string.Join(", ", TeamMembers)}";
    }

    // Новые методы
    public void AddTeamMember(string member)
    {
        if (!TeamMembers.Contains(member))
        {
            TeamMembers.Add(member);
            TeamSize++;
            Console.WriteLine($"Участник '{member}' добавлен в команду '{TeamName}'. Размер: {TeamSize}");
        }
    }

    public void ShowTeam()
    {
        Console.WriteLine($"Команда '{TeamName}' ({TeamSize} чел.): {string.Join(", ", TeamMembers)}");
    }
}

// ============================================================
// 5. ResearchTask — СЛОЖНОЕ + МНОЖЕСТВЕННОЕ наследование (через интерфейсы)
// ============================================================
public class ResearchTask : Task, IReportable, IPrioritizable
{
    public string DataSource { get; set; }
    public string ResearchField { get; set; }
    public DateTime StartDate { get; set; }
    public List<string> Publications { get; set; }

    public ResearchTask(int id, string name, string priority, string assignedTo,
                        string dataSource, string field, DateTime startDate)
        : base(id, name, priority, assignedTo)
    {
        DataSource = dataSource;
        ResearchField = field;
        StartDate = startDate;
        Publications = new List<string>();
    }

    // Переопределение GetTaskDetails (полиморфизм)
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() +
               $" | Источник данных: {DataSource} | Область: {ResearchField} | " +
               $"Начало: {StartDate:dd.MM.yyyy} | Публикаций: {Publications.Count}";
    }

    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[RESEARCH] Исследование '{TaskName}' завершено! " +
                          $"Источник данных: {DataSource}. Область: {ResearchField}.");
    }

    // Реализация IReportable
    public string GenerateReport()
    {
        return $"=== ОТЧЁТ ПО ИССЛЕДОВАНИЮ ===\n" +
               $"Задача: {TaskName}\n" +
               $"Область: {ResearchField}\n" +
               $"Источник: {DataSource}\n" +
               $"Публикации: {string.Join(", ", Publications)}\n" +
               $"Статус: {(IsCompleted ? "Завершено" : "В процессе")}";
    }

    // Реализация IPrioritizable
    public void ChangePriority(string newPriority)
    {
        Console.WriteLine($"Приоритет исследования '{TaskName}' изменён с {Priority} на {newPriority}");
        Priority = newPriority;
    }

    public string GetPriorityLevel()
    {
        return Priority switch
        {
            "High" => "🔴 Критический",
            "Medium" => "🟡 Средний",
            "Low" => "🟢 Низкий",
            _ => "Неизвестно"
        };
    }

    // Новый метод
    public void AddPublication(string title)
    {
        Publications.Add(title);
        Console.WriteLine($"Публикация '{title}' добавлена к исследованию '{TaskName}'.");
    }
}
```

```
Console.WriteLine("========== ДЕМОНСТРАЦИЯ ПРОЕКТА ==========\n");

// --- DelegateTask ---
var dt = new DelegateTask(101, "Настроить сервер", "High", "Иван",
                          DateTime.Now.AddDays(5), "Петров", "IT-отдел", 12);
Console.WriteLine(dt.GetTaskDetails());
dt.MarkAsComplete();
dt.ExtendDeadline(3);
Console.WriteLine($"Просрочена? {dt.IsOverdue()}\n");

// --- TeamTask ---
var tt = new TeamTask(102, "Разработка модуля", "Medium", "Анна",
                      "Backend Team", 3, "Сидоров");
Console.WriteLine(tt.GetTaskDetails());
tt.AddTeamMember("Мария");
tt.ReassignTo("Олег");
tt.ShowTeam();
Console.WriteLine();

// --- ResearchTask ---
var rt = new ResearchTask(103, "Анализ рынка", "Low", "Елена",
                          "Открытые данные Росстата", "Экономика", DateTime.Now);
Console.WriteLine(rt.GetTaskDetails());
rt.AddPublication("Обзор рынка 2025");
Console.WriteLine(rt.GenerateReport());
rt.ChangePriority("High");
Console.WriteLine($"Уровень приоритета: {rt.GetPriorityLevel()}");
rt.MarkAsComplete();
Console.WriteLine();

// --- ПОЛИМОРФИЗМ: список базового типа ---
Console.WriteLine("========== ПОЛИМОРФИЗМ ==========");
List<Task> tasks = new List<Task> { dt, tt, rt };
foreach (var t in tasks)
{
    Console.WriteLine(t.GetTaskDetails());
    Console.WriteLine("---");
}
```
