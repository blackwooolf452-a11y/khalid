#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

**Описание задачи:**

Создан базовый класс `Task` (Задача), представляющий задачу внутри проекта. 
На его основе разработаны три производных класса, демонстрирующих принципы 
наследования и полиморфизма:

1. **DelegateTask** (Делегатская задача) — добавляет атрибут `DueDate` (дата выполнения) 
   и переопределяет метод `MarkAsComplete()`.

2. **TeamTask** (Командная задача) — добавляет атрибут `TeamName` (название команды) 
   и переопределяет метод `ReassignTo()`.

3. **ResearchTask** (Исследовательская задача) — добавляет атрибут `DataSource` 
   (источник данных) и переопределяет метод `GetTaskDetails()`.

**Требования к базовому классу Task:**
- Атрибуты: `TaskId` (ID задачи), `TaskName` (название), `Priority` (приоритет).
- Методы: `MarkAsComplete()`, `GetTaskDetails()`, `ReassignTo()`.

**Демонстрируемые концепции ООП:**
- Инкапсуляция (свойства с get/set)
- Наследование (производные классы от Task)
- Полиморфизм (переопределение методов через virtual/override)
- Взаимодействие объектов между собой

**Дополнительное задание:** 
К существующим классам добавлены конструкторы с использованием свойств (геттеров и 
сеттеров), реализовано взаимодействие объектов между собой.

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// ==================== БАЗОВЫЙ КЛАСС: Task ====================
public class Task
{
    // Свойства (геттеры и сеттеры)
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    
    // Защищённый атрибут — доступен только внутри класса и наследникам
    protected bool IsCompleted { get; set; }

    // Конструктор с использованием свойств
    public Task(int taskId, string taskName, string priority)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        IsCompleted = false;
    }

    // Метод для отметки задачи как выполненной (виртуальный для переопределения)
    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[✓] Задача '{TaskName}' (ID: {TaskId}) отмечена как выполненная.");
    }

    // Метод для получения деталей задачи
    public virtual string GetTaskDetails()
    {
        return $"ID: {TaskId} | Название: {TaskName} | Приоритет: {Priority} | " +
               $"Статус: {(IsCompleted ? "Выполнена" : "В работе")}";
    }

    // Метод для переназначения задачи другому члену команды
    public virtual void ReassignTo(string teamMember)
    {
        Console.WriteLine($"[→] Задача '{TaskName}' переназначена на: {teamMember}");
    }
}

// ==================== КЛАСС 1: DelegateTask ====================
public class DelegateTask : Task
{
    // Дополнительный атрибут
    public DateTime DueDate { get; set; }

    // Конструктор с использованием свойств базового класса
    public DelegateTask(int taskId, string taskName, string priority, DateTime dueDate)
        : base(taskId, taskName, priority)
    {
        DueDate = dueDate;
    }

    // Переопределение метода MarkAsComplete (полиморфизм)
    public override void MarkAsComplete()
    {
        IsCompleted = true;
        Console.WriteLine($"[✓] Делегатская задача '{TaskName}' завершена! " +
                          $"Срок выполнения был: {DueDate:dd.MM.yyyy}");
    }
}

// ==================== КЛАСС 2: TeamTask ====================
public class TeamTask : Task
{
    // Дополнительный атрибут
    public string TeamName { get; set; }

    // Конструктор
    public TeamTask(int taskId, string taskName, string priority, string teamName)
        : base(taskId, taskName, priority)
    {
        TeamName = teamName;
    }

    // Переопределение метода ReassignTo (полиморфизм)
    public override void ReassignTo(string teamMember)
    {
        Console.WriteLine($"[→] Командная задача '{TaskName}' из команды '{TeamName}' " +
                          $"переназначена на: {teamMember}");
    }
}

// ==================== КЛАСС 3: ResearchTask ====================
public class ResearchTask : Task
{
    // Дополнительный атрибут
    public string DataSource { get; set; }

    // Конструктор
    public ResearchTask(int taskId, string taskName, string priority, string dataSource)
        : base(taskId, taskName, priority)
    {
        DataSource = dataSource;
    }

    // Переопределение метода GetTaskDetails (полиморфизм)
    public override string GetTaskDetails()
    {
        return base.GetTaskDetails() + $" | Источник данных: {DataSource}";
    }
}
```

```


// ==================== ПРИМЕР ИСПОЛЬЗОВАНИЯ ====================

Console.WriteLine("========== СОЗДАНИЕ ОБЪЕКТОВ ==========\n");

// Создание экземпляров классов
DelegateTask delegateTask = new DelegateTask(
    101, "Подготовить отчёт", "High", new DateTime(2026, 10, 15));

TeamTask teamTask = new TeamTask(
    102, "Разработать модуль", "Medium", "Backend Team");

ResearchTask researchTask = new ResearchTask(
    103, "Анализ рынка", "Low", "Открытые данные Росстата");

Console.WriteLine("========== ВСЕ ЗАДАЧИ ПРОЕКТА ==========\n");

// Полиморфизм — массив объектов базового типа
Task[] allTasks = { delegateTask, teamTask, researchTask };
foreach (Task t in allTasks)
{
    Console.WriteLine(t.GetTaskDetails());
}

Console.WriteLine("\n========== ДЕЙСТВИЯ С ЗАДАЧАМИ ==========\n");

// Вызов переопределённых методов
delegateTask.MarkAsComplete();
teamTask.ReassignTo("Иван Петров");

Console.WriteLine("\n========== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ==========\n");

// Взаимодействие между объектами: TeamTask взаимодействует с DelegateTask
Console.WriteLine($"Задача '{teamTask.TaskName}' взаимодействует с задачей '{delegateTask.TaskName}':");
teamTask.ReassignTo($"Команда задачи: {delegateTask.TaskName}");

Console.WriteLine("\n========== ИТОГОВЫЕ ДЕТАЛИ ==========\n");
Console.WriteLine(delegateTask.GetTaskDetails());
Console.WriteLine(teamTask.GetTaskDetails());
Console.WriteLine(researchTask.GetTaskDetails());
```
