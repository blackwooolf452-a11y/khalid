<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

```csharp
using System;
using System.Collections.Generic;

// Определение базового класса Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

// Определение производного класса Car
public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

// Определение производного класса Motorcycle
public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Определение генерического класса VehicleCollection<T>
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
    }

    public void Remove(T vehicle)
    {
        _vehicles.Remove(vehicle);
    }

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive(); // Вызов метода Drive() для каждого транспортного средства
        }
    }
}


VehicleCollection<Car> cars = new VehicleCollection<Car>();
cars.Add(new Car());
cars.Add(new Motorcycle()); // Это возможно благодаря ограничению "where T : Vehicle"

cars.DisplayVehicles(); // Выводит: "Машина едет." и "Мотоцикл едет."


```

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

### Заключение

Генерические классы в C# предоставляют мощный и гибкий способ создания компонентов, которые могут работать с различными типами данных, обеспечивая при этом строгую типизацию и безопасность типов. Это делает их идеальным выбором для многих задач, требующих повторного использования кода с высокой степенью абстракции.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке:</h4>

----

````
using System;
using System.Collections.Generic;

// Базовый класс Vehicle
public class Vehicle
{
    public virtual void Drive()
    {
        Console.WriteLine("Транспортное средство находится в движении.");
    }
}

public class Car : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Машина едет.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Drive()
    {
        Console.WriteLine("Мотоцикл едет.");
    }
}

// Генерический класс с ограничением where T : Vehicle
public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();

    public void Add(T vehicle) => _vehicles.Add(vehicle);
    public void Remove(T vehicle) => _vehicles.Remove(vehicle);

    public void DisplayVehicles()
    {
        foreach (var vehicle in _vehicles)
            vehicle.Drive();
    }
}
````

````

// Использование
VehicleCollection<Vehicle> vehicles = new VehicleCollection<Vehicle>();
vehicles.Add(new Car());
vehicles.Add(new Motorcycle());
vehicles.Add(new Vehicle());

vehicles.DisplayVehicles();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также с пременением Generic классов.

````
using System;
using System.Collections.Generic;

// ================== Базовый класс Animal ==================
public class Animal
{
    // Свойства с геттерами/сеттерами
    public string Name { get; set; }
    public int Age { get; set; }
    public string Species { get; protected set; }

    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        Species = species;
    }

    // Виртуальный метод для полиморфизма
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} ({Species}) издаёт какой-то звук.");
    }

    // Виртуальный метод с параметром
    public virtual void Move(int distance)
    {
        Console.WriteLine($"{Name} перемещается на {distance} метров.");
    }

    public void Info()
    {
        Console.WriteLine($"{Name} | {Species} | {Age} лет");
    }
}

// ================== Производные классы ==================
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed) : base(name, age, "Собака")
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} ({Breed}) говорит: Гав-гав!");
    }

    public override void Move(int distance)
    {
        Console.WriteLine($"{Name} бежит на {distance} метров, виляя хвостом.");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, bool isIndoor) : base(name, age, "Кошка")
    {
        IsIndoor = isIndoor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public override void Move(int distance)
    {
        string place = IsIndoor ? "по квартире" : "по улице";
        Console.WriteLine($"{Name} крадётся {place} на {distance} метров.");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }

    public Bird(string name, int age, double wingspan) : base(name, age, "Птица")
    {
        Wingspan = wingspan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поёт: Чирик-чирик!");
    }

    public override void Move(int distance)
    {
        Console.WriteLine($"{Name} летит на {distance} метров (размах крыльев {Wingspan} м).");
    }
}

public class Fish : Animal
{
    public bool IsFreshwater { get; set; }

    public Fish(string name, int age, bool isFreshwater) : base(name, age, "Рыба")
    {
        IsFreshwater = isFreshwater;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} молчит и пускает пузыри...");
    }

    public override void Move(int distance)
    {
        string water = IsFreshwater ? "в пресной воде" : "в солёной воде";
        Console.WriteLine($"{Name} плывёт {distance} метров {water}.");
    }
}

// ================== GENERIC-класс: коллекция животных ==================
// Ограничение where T : Animal позволяет быть уверенным,
// что у элементов доступны Name, MakeSound(), Move() и т.д.
public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public int Count => _animals.Count;

    public void Add(T animal) => _animals.Add(animal);

    public bool Remove(T animal) => _animals.Remove(animal);

    // Демонстрация полиморфизма
    public void MakeAllSounds()
    {
        foreach (var a in _animals)
            a.MakeSound();
    }

    public void MoveAll(int distance)
    {
        foreach (var a in _animals)
            a.Move(distance);
    }

    public void ShowAll()
    {
        foreach (var a in _animals)
            a.Info();
    }

    // Поиск самого старого животного
    public T Oldest()
    {
        T oldest = null;
        foreach (var a in _animals)
            if (oldest == null || a.Age > oldest.Age)
                oldest = a;
        return oldest;
    }
}
````

````


// ================== Демонстрация ==================

// 1) Общая коллекция всех животных — типизирована базовым классом
var zoo = new AnimalCollection<Animal>();
zoo.Add(new Dog("Рекс", 3, "Лабрадор"));
zoo.Add(new Cat("Мурка", 2, true));
zoo.Add(new Bird("Кеша", 1, 0.3));
zoo.Add(new Fish("Немо", 1, false));

Console.WriteLine("=== Все животные ===");
zoo.ShowAll();

Console.WriteLine("\n=== Звуки ===");
zoo.MakeAllSounds();

Console.WriteLine("\n=== Движение ===");
zoo.MoveAll(30);

Console.WriteLine($"\nВсего животных: {zoo.Count}");
Console.WriteLine($"Самое старое: {zoo.Oldest().Name}");

// 2) Специализированная коллекция — только собаки
var kennel = new AnimalCollection<Dog>();
kennel.Add(new Dog("Шарик", 5, "Дворняга"));
kennel.Add(new Dog("Бим", 2, "Боксёр"));
// kennel.Add(new Cat(...)); // ❌ Ошибка компиляции — типобезопасность!
// kennel.Add(new Animal(...)); // ❌ Ошибка компиляции — T должен быть Dog

Console.WriteLine("\n=== Питомник собак ===");
kennel.ShowAll();
kennel.MakeAllSounds();

// 3) Коллекция только птиц
var aviary = new AnimalCollection<Bird>();
aviary.Add(new Bird("Попугай", 1, 0.2));
aviary.Add(new Bird("Орёл", 4, 2.0));

Console.WriteLine("\n=== Вольер птиц ===");
aviary.ShowAll();
aviary.MoveAll(100);
````
